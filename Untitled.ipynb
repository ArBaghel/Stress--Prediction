```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle, os, gc, warnings, multiprocessing
warnings.filterwarnings('ignore')

from scipy.signal import butter, filtfilt
from sklearn.model_selection import train_test_split
from sklearn.metrics import (classification_report, confusion_matrix,
                             roc_auc_score, roc_curve, recall_score,
                             f1_score, precision_score)
from imblearn.over_sampling import SMOTE

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Conv1D, BatchNormalization,
                                     MaxPooling1D, Dropout, Bidirectional,
                                     LSTM, Dense, Multiply, Flatten,
                                     Activation)
from tensorflow.keras.callbacks import (EarlyStopping, ReduceLROnPlateau,
                                        ModelCheckpoint)
from tensorflow.keras.optimizers import Adam

# CPU optimization
cores = multiprocessing.cpu_count()
tf.config.threading.set_inter_op_parallelism_threads(cores)
tf.config.threading.set_intra_op_parallelism_threads(cores)

np.random.seed(42)
tf.random.set_seed(42)

# Paths
WESAD_DIR  = r"D:\Projects\Stress_ecg_project\data\WESAD"
MODELS_DIR = r"D:\Projects\Stress_ecg_project\models"
PLOTS_DIR  = r"D:\Projects\Stress_ecg_project\plots"

FS          = 700
WINDOW_SIZE = 250
STEP_SIZE   = 125

print(f"TensorFlow  : {tf.__version__}")
print(f"NumPy       : {np.__version__}")
print(f"CPU cores   : {cores}")
print(f"WESAD path  : {WESAD_DIR}")
print("All imports done.")
```

```python
SUBJECTS = ['S2', 'S3', 'S4', 'S5','S6','S7','S8','S9','S10','S11','S13','S13','S14','S15','S16','S17']

def load_subject(sid):
    path = os.path.join(WESAD_DIR, sid, f"{sid}.pkl")

    if not os.path.exists(path):
        for root, dirs, files in os.walk(WESAD_DIR):
            if f"{sid}.pkl" in files:
                path = os.path.join(root, f"{sid}.pkl")
                break

    with open(path, 'rb') as f:
        data = pickle.load(f, encoding='latin1')

    ecg    = data['signal']['chest']['ECG'].flatten().astype(np.float32)
    labels = data['label'].flatten()
    del data; gc.collect()

    mask   = (labels == 1) | (labels == 2)
    ecg    = ecg[mask]
    labels = (labels[mask] == 2).astype(np.int8)

    return ecg, labels


all_ecg, all_labels = [], []
print("Loading subjects...\n")

for sid in SUBJECTS:
    ecg, labels = load_subject(sid)
    all_ecg.append(ecg)
    all_labels.append(labels)
    print(f"  {sid} | samples: {len(ecg):,} | "
          f"stress: {labels.sum():,} | "
          f"no-stress: {(labels==0).sum():,}")

ecg_full    = np.concatenate(all_ecg)
labels_full = np.concatenate(all_labels)
del all_ecg, all_labels; gc.collect()

print(f"\nDataset loaded.")
print(f"  Total samples : {len(ecg_full):,}  ({len(ecg_full)/FS/60:.1f} min)")
print(f"  Stress        : {labels_full.sum():,}  ({labels_full.mean()*100:.1f}%)")
print(f"  No-stress     : {(labels_full==0).sum():,}  ({(1-labels_full.mean())*100:.1f}%)")
print(f"  RAM used      : {ecg_full.nbytes/1024/1024:.1f} MB")
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
counts  = [(labels_full==0).sum(), (labels_full==1).sum()]
bars    = ax.bar(['No-Stress', 'Stress'], counts,
                 color=['steelblue', 'tomato'],
                 edgecolor='white', width=0.5)
for bar, val in zip(bars, counts):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + max(counts)*0.01,
            f'{val:,}', ha='center', fontweight='bold')
ax.set_title('Class Distribution', fontweight='bold')
ax.set_ylabel('Samples')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'eda_class_distribution.png'), dpi=150)
plt.show()

# ── No-Stress ECG ─────────────────────────────────────────────
no_stress_start = np.where(labels_full == 0)[0][0]
window          = 5 * FS
t               = np.arange(window) / FS

fig, ax = plt.subplots(figsize=(14, 3))
ax.plot(t, ecg_full[no_stress_start:no_stress_start+window],
        color='steelblue', linewidth=0.8)
ax.set_title('No-Stress ECG — 5 seconds', fontweight='bold')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'eda_no_stress_ecg.png'), dpi=150)
plt.show()

# ── Stress ECG ────────────────────────────────────────────────
stress_start = np.where(labels_full == 1)[0][0]

fig, ax = plt.subplots(figsize=(14, 3))
ax.plot(t, ecg_full[stress_start:stress_start+window],
        color='tomato', linewidth=0.8)
ax.set_title('Stress ECG — 5 seconds', fontweight='bold')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'eda_stress_ecg.png'), dpi=150)
plt.show()

# ── Signal Statistics ─────────────────────────────────────────
print("Signal Statistics:")
print(f"  Mean      : {ecg_full.mean():.6f}")
print(f"  Std       : {ecg_full.std():.6f}")
print(f"  Min       : {ecg_full.min():.6f}")
print(f"  Max       : {ecg_full.max():.6f}")
print(f"  Imbalance : {(labels_full==0).sum()/(labels_full==1).sum():.2f}x")
```

```python
from scipy.signal import butter, filtfilt

# ── 1. Bandpass Filter ────────────────────────────────────────
def bandpass_filter(ecg, lowcut=0.5, highcut=40.0, fs=700, order=4):
    nyq  = fs / 2
    b, a = butter(order, [lowcut/nyq, highcut/nyq], btype='band')
    return filtfilt(b, a, ecg).astype(np.float32)

print("Step 1: Bandpass filtering...")
ecg_filtered = bandpass_filter(ecg_full)
print(f"        Done.")

# Plot 1 — Raw ECG
PLOT_SAMPLES = 3 * FS
t = np.arange(PLOT_SAMPLES) / FS

fig, ax = plt.subplots(figsize=(14, 3))
ax.plot(t, ecg_full[:PLOT_SAMPLES], color='gray', linewidth=0.8)
ax.set_title('Raw ECG before filtering — 3 seconds', fontweight='bold')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'ecg_raw.png'), dpi=150)
plt.show()

# Plot 2 — Filtered ECG
fig, ax = plt.subplots(figsize=(14, 3))
ax.plot(t, ecg_filtered[:PLOT_SAMPLES], color='steelblue', linewidth=0.8)
ax.set_title('Filtered ECG after bandpass 0.5-40 Hz — 3 seconds',
             fontweight='bold')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'ecg_filtered.png'), dpi=150)
plt.show()
print("        Raw vs filtered plots saved.")

# ── 2. Windowing ──────────────────────────────────────────────
def create_windows(ecg, labels, window_size=250, step=125):
    X, y = [], []
    for start in range(0, len(ecg) - window_size, step):
        end = start + window_size
        X.append(ecg[start:end])
        y.append(1 if labels[start:end].mean() > 0.5 else 0)
    return np.array(X, dtype=np.float32), np.array(y, dtype=np.int8)

print("Step 2: Creating windows...")
X, y = create_windows(ecg_filtered, labels_full)
print(f"        {len(X):,} total windows created")

# ── 3. Normalize ──────────────────────────────────────────────
print("Step 3: Normalizing...")
mean = X.mean(axis=1, keepdims=True)
std  = X.std(axis=1,  keepdims=True)
X    = (X - mean) / (std + 1e-8)
print(f"        Done. Sample mean: {X[0].mean():.4f}  std: {X[0].std():.4f}")

del ecg_filtered, ecg_full; gc.collect()

# ── 4. Smart Subsample ────────────────────────────────────────
print("Step 4: Smart subsampling...")
MAX_PER_CLASS = 25000

stress_idx    = np.where(y == 1)[0]
no_stress_idx = np.where(y == 0)[0]

np.random.seed(42)
stress_sample    = np.random.choice(stress_idx,
                   min(MAX_PER_CLASS, len(stress_idx)), replace=False)
no_stress_sample = np.random.choice(no_stress_idx,
                   min(MAX_PER_CLASS, len(no_stress_idx)), replace=False)

selected_idx = np.concatenate([stress_sample, no_stress_sample])
np.random.shuffle(selected_idx)

X = X[selected_idx]
y = y[selected_idx]
print(f"        Subsampled to {len(X):,} windows")
print(f"        Stress    : {(y==1).sum():,}")
print(f"        No-stress : {(y==0).sum():,}")

# Plot 3 — Class balance
fig, ax = plt.subplots(figsize=(6, 4))
counts = [(y==0).sum(), (y==1).sum()]
bars   = ax.bar(['No-Stress', 'Stress'], counts,
                color=['steelblue', 'tomato'],
                edgecolor='white', width=0.5)
for bar, val in zip(bars, counts):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + max(counts)*0.01,
            f'{val:,}', ha='center', fontweight='bold')
ax.set_title('Class Balance after Subsampling', fontweight='bold')
ax.set_ylabel('Windows')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'class_balance.png'), dpi=150)
plt.show()

# ── 5. Split 70/15/15 ─────────────────────────────────────────
print("Step 5: Splitting 70 / 15 / 15...")
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=0.176, random_state=42, stratify=y_temp)
print(f"        Train : {len(X_train):,}")
print(f"        Val   : {len(X_val):,}")
print(f"        Test  : {len(X_test):,}")

# ── 6. Add channel dim ────────────────────────────────────────
X_train_bal = X_train[..., np.newaxis]
X_val       = X_val[..., np.newaxis]
X_test      = X_test[..., np.newaxis]
y_train_bal = y_train

# Plot 4 — 20 sample windows
np.random.seed(42)
sample_idx = np.random.choice(len(X_train_bal), 20, replace=False)

fig, axes = plt.subplots(4, 5, figsize=(16, 10))
axes = axes.flatten()

for i, idx in enumerate(sample_idx):
    window = X_train_bal[idx].flatten()
    label  = y_train_bal[idx]
    color  = 'steelblue' if label == 0 else 'tomato'

    axes[i].plot(window, color=color, linewidth=0.8)
    axes[i].set_title(
        f"Sample {i+1}\n{'No-Stress' if label==0 else 'Stress'}",
        fontsize=8, fontweight='bold')
    axes[i].set_xticks([])
    axes[i].set_yticks([])
    for spine in ['top', 'right']:
        axes[i].spines[spine].set_visible(False)
    axes[i].spines['left'].set_color(color)
    axes[i].spines['bottom'].set_color(color)

plt.suptitle(
    f"20 Random Windows after Preprocessing  |  "
    f"Blue = No-Stress   Red = Stress  |  "
    f"Total: {len(X_train_bal):,}",
    fontsize=10, fontweight='bold'
)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'sample_windows.png'),
            dpi=150, bbox_inches='tight')
plt.show()

print(f"\nPreprocessing complete.")
print(f"  X_train : {X_train_bal.shape}")
print(f"  X_val   : {X_val.shape}")
print(f"  X_test  : {X_test.shape}")
print(f"  Plots saved to : {PLOTS_DIR}")
```

```python
from tensorflow.keras.regularizers import l2
def attention_block(inputs):
    """
    Attention mechanism — learns which ECG timesteps
    matter most for stress detection.
    Sigmoid scores each timestep independently.
    """
    score   = Dense(1, activation='tanh')(inputs)
    score   = Activation('sigmoid')(score)
    context = Multiply()([inputs, score])
    return context, score


def build_model(input_shape=(250, 1)):
    """
    CNN-BiGRU + Attention model.

    Flow:
      Input ECG (250 samples)
        -> CNN (local pattern detection)
        -> BiGRU (temporal dependencies)
        -> Attention (focus on important beats)
        -> Dense (classification)
        -> Output (0=No-Stress, 1=Stress)
    """
    inputs = Input(shape=input_shape, name='ecg_input')

    # ── CNN Block 1 ───────────────────────────────────────────
    # Detects local ECG features like QRS peaks
    x = Conv1D(64, kernel_size=5, padding='same',
               activation='relu', name='conv1')(inputs)
    x = BatchNormalization(name='bn1')(x)

    # ── CNN Block 2 ───────────────────────────────────────────
    # Learns combinations of local features
    x = Conv1D(128, kernel_size=5, padding='same',
               activation='relu', name='conv2')(x)
    x = BatchNormalization(name='bn2')(x)
    x = MaxPooling1D(pool_size=2, name='pool')(x)
    x = Dropout(0.3, name='drop1')(x)

    # ── BiGRU Block 1 ─────────────────────────────────────────
    # Reads ECG rhythm forward and backward
    # GRU is 3x faster than LSTM on CPU
    x = Bidirectional(
            tf.keras.layers.GRU(64, return_sequences=True,
                                dropout=0.2),
            name='bigru1')(x)

    # ── BiGRU Block 2 ─────────────────────────────────────────
    x = Bidirectional(
            tf.keras.layers.GRU(32, return_sequences=True,
                                dropout=0.2),
            name='bigru2')(x)

    # ── Attention ─────────────────────────────────────────────
    # Highlights stress-relevant ECG segments
    x, _ = attention_block(x)
    x     = Flatten(name='flatten')(x)

    # ── Dense Classification Head ──────────────────────────────
    x      = Dense(64, activation='relu',
               kernel_regularizer=l2(0.001),
               name='dense1')(x)
    x      = Dropout(0.4, name='drop2')(x)
    x      = Dense(32, activation='relu',
               kernel_regularizer=l2(0.001),
               name='dense2')(x)
    x      = Dropout(0.3, name='drop3')(x)
    output = Dense(1, activation='sigmoid', name='output')(x)

    return Model(inputs=inputs, outputs=output,
                 name='CNN_BiGRU_Attention')


# Build model
model = build_model(input_shape=(250, 1))

# Compile
model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='binary_crossentropy',
    metrics=[
        'accuracy',
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall'),
        tf.keras.metrics.AUC(name='auc')
    ]
)

model.summary()

print(f"\nModel ready.")
print(f"  Parameters    : {model.count_params():,}")
print(f"  Input shape   : (250, 1)")
print(f"  Output shape  : (1,)  — 0=No-Stress  1=Stress")
```

```python
# No subsampling needed — already done in Cell 4
print(f"Train : {len(X_train_bal):,} | Val : {len(X_val):,}")

# Check CPU cores
import multiprocessing
cores = multiprocessing.cpu_count()
print(f"CPU cores available : {cores}")

callbacks = [
    EarlyStopping(monitor='val_loss', patience=7,
                  restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                      patience=4, min_lr=1e-6, verbose=1),
    ModelCheckpoint(
        filepath=os.path.join(MODELS_DIR, 'best_model.keras'),
        monitor='val_auc', save_best_only=True,
        mode='max', verbose=1)
]

print("\nTraining started...")
print(f"  Train   : {len(X_train_bal):,} windows")
print(f"  Val     : {len(X_val):,} windows")
print(f"  Batch   : 256")
print(f"  Epochs  : up to 50")
print(f"  Device  : CPU ({cores} cores)\n")

history = model.fit(
    X_train_bal, y_train_bal,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=256,
    callbacks=callbacks,
    verbose=1
)

best_epoch = np.argmax(history.history['val_auc']) + 1
print(f"\nTraining complete.")
print(f"  Best epoch   : {best_epoch}")
print(f"  Best val_acc : {max(history.history['val_accuracy'])*100:.2f}%")
print(f"  Best val_auc : {max(history.history['val_auc']):.4f}")
print(f"  Best val_rec : {max(history.history['val_recall']):.4f}")
```

```python
# ── Confidence sanity check ───────────────────────────────────
print("\nRunning confidence check...")
sample_prob = model.predict(X_val[:500], verbose=0).flatten()

print(f"  Min confidence  : {sample_prob.min()*100:.1f}%")
print(f"  Max confidence  : {sample_prob.max()*100:.1f}%")
print(f"  Mean confidence : {sample_prob.mean()*100:.1f}%")
print(f"  Std confidence  : {sample_prob.std()*100:.1f}%")

if sample_prob.std() > 0.15:
    print("\n  Model is predicting with real confidence.")
    print("  Spread is healthy — not guessing.")
else:
    print("\n  Warning — model may still be guessing near 50%.")
    print("  Std too low. Try retraining with more epochs.")

# Show distribution
below_30  = (sample_prob < 0.30).sum()
between   = ((sample_prob >= 0.30) & (sample_prob <= 0.70)).sum()
above_70  = (sample_prob > 0.70).sum()

print(f"\n  Confidence distribution (500 samples):")
print(f"    Below 30% (confident No-Stress) : {below_30}")
print(f"    30% to 70% (uncertain zone)     : {between}")
print(f"    Above 70% (confident Stress)    : {above_70}")

if between > 400:
    print("\n  Too many uncertain predictions.")
    print("  Model needs more training data or epochs.")
else:
    print("\n  Good — most predictions are outside uncertain zone.")
```

```python

from sklearn.metrics import (classification_report, confusion_matrix,
                             roc_auc_score, roc_curve,
                             recall_score, f1_score, precision_score)

y_pred_prob = model.predict(X_test, verbose=0).flatten()
y_pred      = (y_pred_prob > 0.5).astype(int)

acc         = (y_pred == y_test).mean()
auc         = roc_auc_score(y_test, y_pred_prob)
recall      = recall_score(y_test, y_pred)
precision   = precision_score(y_test, y_pred)
f1          = f1_score(y_test, y_pred)
tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
specificity = tn / (tn + fp)

print("=" * 55)
print("         FINAL TEST RESULTS")
print("=" * 55)
print(classification_report(y_test, y_pred,
      target_names=['No-Stress', 'Stress']))
print(f"  Accuracy    : {acc*100:.2f}%   (target >= 92%)")
print(f"  F1-Score    : {f1:.4f}    (target >= 0.90)")
print(f"  Sensitivity : {recall*100:.2f}%   (target >= 94%)")
print(f"  Specificity : {specificity*100:.2f}%   (target >= 90%)")
print(f"  Precision   : {precision*100:.2f}%")
print(f"  AUC-ROC     : {auc:.4f}    (target >= 0.95)")
print("=" * 55)

# Target check
print("\n  Target Summary:")
checks = {
    "Accuracy    >= 92%" : acc >= 0.92,
    "F1          >= 0.90": f1  >= 0.90,
    "Sensitivity >= 94%" : recall >= 0.94,
    "Specificity >= 90%" : specificity >= 0.90,
    "AUC-ROC     >= 0.95": auc >= 0.95,
}
for metric, passed in checks.items():
    print(f"  {'PASSED' if passed else 'CLOSE '} — {metric}")

# Plots
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Confusion matrix
sns.heatmap(confusion_matrix(y_test, y_pred),
            annot=True, fmt='d', cmap='Blues',
            xticklabels=['No-Stress', 'Stress'],
            yticklabels=['No-Stress', 'Stress'],
            ax=axes[0])
axes[0].set_title('Confusion Matrix', fontweight='bold')
axes[0].set_ylabel('Actual')
axes[0].set_xlabel('Predicted')
axes[0].text(0.5, -0.15,
             f"TP={tp}  TN={tn}  FP={fp}  FN={fn}",
             transform=axes[0].transAxes,
             ha='center', fontsize=9, color='gray')

# ROC curve
fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
axes[1].plot(fpr, tpr, color='steelblue',
             lw=2, label=f'AUC = {auc:.4f}')
axes[1].plot([0,1],[0,1], 'k--', lw=1, label='Random')
axes[1].fill_between(fpr, tpr, alpha=0.1, color='steelblue')
axes[1].scatter([1-specificity], [recall],
                color='tomato', s=100, zorder=5,
                label='Operating point')
axes[1].set_title('ROC Curve', fontweight='bold')
axes[1].set_xlabel('False Positive Rate')
axes[1].set_ylabel('True Positive Rate')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Training history
axes[2].plot(history.history['accuracy'],
             color='steelblue', lw=2, label='Train')
axes[2].plot(history.history['val_accuracy'],
             color='tomato', lw=2, label='Val')
axes[2].axhline(y=0.92, color='green', lw=1,
                linestyle='--', label='Target 92%')
axes[2].set_title('Training History', fontweight='bold')
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('Accuracy')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.suptitle('CNN-BiGRU + Attention — Final Results',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'final_results.png'),
            dpi=150, bbox_inches='tight')
plt.show()
print(f"\nPlots saved to {PLOTS_DIR}")
```
